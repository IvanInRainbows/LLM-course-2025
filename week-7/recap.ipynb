# Prompting

It's esentially how we communicate with the model. By prompting we can find some emerging capability of the models. In some sense the model is only a part of the software because then theres the part of how the queries are processed or all the things that are happening between the user's input until that input reaches the LLM.

# Finetuning

The important distinction is the traditional ML models and the foundational models. In the first ones you have to nail the data districution and a single model is trained for each task. The second one is trained for general tasks with tons amount of data and you can use it for prompting, but they are not great as such, so they have to be **fine tuned**. This can be done by prompting, finetuning, or RAG.

SFT: is composed of pairs of input-output with the expected/desired output

Instruction tuning: specifically for LLMs and for chat-like instructions.

In HF they usually have the name *base* for the first one and *instruction* for the second one.

RLHF: combines the first type with reinforcment learning and requires human annotator, thus is costly and more complex. Designing how the data is annotated is not an easy process.

# Evaluation

Benchmarking is form evaluating the performance using standardized tasks and datasets. This are standardized tasks often used by academics to evaluate the models. Some are MMLU, GLUE, HellaSwag and ELOQUENT.

Validations is use-case specific, evaluates whether the model works well in a specific dataset. This dataset may be in a specific topic or belong to a specific domain.

# RAG

A variant of prompting that gives the chance to adapt to changing data, because data changes constantly. RAG makes it so the model does not need to be re-trained by injecting information from, for example, a vector database so it has more updated context and data. The information does non need to be in vector form, but makes it easier and smoother. The process uses vector search, which is based one arithmetic operations in a vector space, which may be LNN, ANN and, I guess, cosine similarity. The search method is not strict, several search methods can be combined by applying weights to them.

## Agentic RAG

It's supposed to have some level of autonomy. The agent is the LLM and has a defined role and task. The content of the chat can be stored (temporarily or not) to use that information. There's a layer that decides which model or module to load for the task.

## Algorithms

Although we may not need to deal with this, the algorithm used impacts the performance of the system, some work better on large amount of data, others work better with less data.

# Use cases and applications for LLM

LLMs are evolving onto multimodality, managing language but also other kind of inputs such as images, video, audio, etc. The method for RAG of skipping OCR and recognizing a PDF as an image allows for faster processing, specially for images into the PDF. These multimodal LLMs also work by encoding each frame from a video as a separate image, but these images must also encode the timestamp, the sequence order is very important.

Use case are varied like chat systems for support or assistance for specific domains. This can be implemented to partially automate stuff or to retrieve more information if there's a issue for a support team. These models can also create synthetic data to get a starting point for a system, which should be replaced lated by proper quality data.
