# RAG

A two phase process based on the retrieval of a document on a database, an evidence for the query, and you give it to the model as context. There as a retriever of documents and a generatos. The first one is non parametric, generally returns the top k. The accuracy depends on the correctness of the documents.

Agentic RAG involves decision making and such, mimicking humans, memory... The memory also works by remembering (storing) details about the user and the past queries though a set of parameters that are heltpful to adapt the info to the user. This has access to APIs, can pre-process, can validate retrieved information.

## Vector search

USing the encoding part of an LLM to get the vectors. It works on nearest neighbor, which is based on euclidean distance between two points (vectors). You can just use the distance but this gets crazy on a lot of data, to solve this it's also possible to divide the data into clusters. The most used algorithm is Hierarchical Navigable small Worlds.
