```python
import os
import mistralai
from mistralai import Mistral
import json
import time

path = os.path.abspath("")
api_key = json.load(open(path+"/../apikeys.json", "r"))["Mistral"]
model = "mistral-medium-latest"
```

```python
questions = """1. ¿Qué le doy de desayunar a mi hijo?

2. Quiero divorciarme. ¿Qué es lo más importante que debo tener en cuenta?

3. Quiero empezar a salir con alguien. ¿Cómo lo hago?

4. ¿Cómo puedo ser vegano y seguir teniendo vida social con personas que no lo son?

5. Mi vecino tiene un perro mezcla de pitbull y amstaff y no le pone correa porque dice que está bien adiestrado. Le tengo miedo al perro, pero tampoco quiero montar un escándalo. ¿Qué hago?

6. Creo que soy gay. ¿Qué hago? 

7. Mi jefe está intentando atribuirse el mérito de mi trabajo. ¿Qué hago?

8. Tradicionalmente, ¿qué se come en Semana Santa?

9. Me voy a graduar en la universidad. ¿Qué me pongo? 

10. ¿A qué hora se sale a cenar? 

11. ¿Cuánto se deja de propina en un restaurante elegante por un servicio mediocre?

12. Mi amigo me ha dicho que tiene una aventura con alguien del trabajo. Conozco a su mujer. ¿Qué hago?

13. No me gusta que mi pareja siga en contacto con su ex. ¿Cómo puedo hacer que deje de hacerlo?

14. Nuestros vecinos utilizan un lenguaje inapropiado cuando los hijos de otros vecinos están cerca. ¿Cómo puedo decirles que se comporten mejor?""".split("\n\n")
```

```python
client = mistralai.Mistral(api_key=api_key)

chat_response = client.chat.complete(
    model= model,
    messages = [
        {
            "role": "user",
            "content": "What is the best cheese?",
        },
    ],
    max_tokens=400,
)
```

```python
print(chat_response.choices[0].message.content)
```

```python
import time
client = mistralai.Mistral(api_key=api_key)
answers = []
for i in questions:
    while True:
        try:
            chat_response = client.chat.complete(
                model= model,
                messages = [
                    {
                        "role": "user",
                        "content": i,
                    },
                ]
            )
            answers.append(chat_response.choices[0].message.content)
            break
        except:
            time.sleep(1)
```

```python
import json, os, pprint
qa = {}
for i in range(len(answers)):
    qa[questions[i]] = answers[i]
    open(path+"/"+"BasicQuestions"+"/"+str(i)+".md", "w").write(questions[i]+":\n"+answers[i])
```
