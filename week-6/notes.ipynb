# Use cases and applications for LLMs

Multimodal LLMs can also manage images, video or audio.

Contrastive learning, links similar concepts together and dissimilates opposite or unrelated concepts. Used for text to image. Some models use image recognition for RAG instead of extracting the text with OCR.

When masking tokens in a transoformer the model doesn't really know the position of the token that has to be predicted, and when doing text ML that's fine. But that's not so fine for video and audio processing.

RAG can be done the opposite way by creating questions from a document, this is GAR.
